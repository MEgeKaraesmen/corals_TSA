```python
"""
wavelet_analysis.py
===================
Continuous Wavelet Transform (CWT) analysis of coral isotope records,
ENSO index, and tide gauge data.

Method: Morlet(6) wavelet following Torrence & Compo (1998).
Significance: 95% red-noise (AR1) threshold.

All datasets use identical wavelet parameters so that significance
contours can be directly compared in overlay plots.

Data loading is handled by 20260223_read_data.py which must be in
the same directory (or on the path).  Run with:
    %run 20260223_read_data.py          # in Jupyter / IPython
    exec(open('20260223_read_data.py').read())  # in plain Python

After that script runs the following objects are available:
    df_sorted       – raw coral data sorted by age_final
    df_resampled    – monthly-resampled coral data
    coefs_centered  – rolling OLS slope/intercept
    dark_bands      – dark-band start/end pairs

Author: <your name>
Date:   2026-02
"""

# =============================================================================
# 0. IMPORTS
# =============================================================================
import matplotlib
import matplotlib.patches as mpatches
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import pycwt as wavelet
from pycwt.helpers import find
from scipy import signal


# =============================================================================
# 1. GLOBAL WAVELET PARAMETERS  (single source of truth)
# =============================================================================
MOTHER   = wavelet.Morlet(6)
DT       = 1 / 12          # sampling interval – 1 month in decimal years
DJ       = 1 / 24          # 24 sub-octaves per octave (fine resolution)
S0       = 1 * DT          # smallest scale  (~1 month)
J        = int(7 / DJ)     # covers up to 2^7 ≈ 128 months ≈ 10.7 years
SIG_LVL  = 0.95            # significance level
SCALE_AVG_BAND = (0.5, 4)  # years – scale-averaged power band (all datasets)


# =============================================================================
# 2. CORE WAVELET FUNCTIONS
# =============================================================================

def remove_trend(data):
    """Linear detrend via scipy (returns detrended array)."""
    return signal.detrend(data)


def run_cwt(data, dt=DT, dj=DJ, s0=S0, j=J, mother=MOTHER):
    """
    Run CWT and return a dict of all products needed for plotting
    and significance testing.
    """
    n           = len(data)
    dat_notrend = remove_trend(data)
    std         = dat_notrend.std()
    var         = std ** 2
    dat_norm    = dat_notrend / std

    alpha, _, _ = wavelet.ar1(data)   # AR1 lag-1 for red noise background

    wave, scales, freqs, coi, fft, fftfreqs = wavelet.cwt(
        dat_norm, dt, dj, s0, j, mother
    )
    iwave  = wavelet.icwt(wave, scales, dt, dj, mother) * std
    power  = (np.abs(wave)) ** 2
    power /= scales[:, None]          # scale-normalised (Torrence & Compo eq. 18)

    fft_power = np.abs(fft) ** 2
    period    = 1 / freqs

    # --- pointwise 95% significance vs red noise
    signif, fft_theor = wavelet.significance(
        1.0, dt, scales, 0, alpha,
        significance_level=SIG_LVL, wavelet=mother
    )
    sig95 = np.ones([1, n]) * signif[:, None]
    sig95 = power / sig95             # >1 means significant

    # --- global wavelet spectrum
    glbl_power  = power.mean(axis=1)
    dof         = n - scales
    glbl_signif, _ = wavelet.significance(
        var, dt, scales, 1, alpha,
        significance_level=SIG_LVL, dof=dof, wavelet=mother
    )

    # --- scale-averaged power over SCALE_AVG_BAND
    sel         = find((period >= SCALE_AVG_BAND[0]) & (period < SCALE_AVG_BAND[1]))
    Cdelta      = mother.cdelta
    scale_avg   = (scales * np.ones((n, 1))).T
    scale_avg   = power / scale_avg
    scale_avg   = var * dj * dt / Cdelta * scale_avg[sel, :].sum(axis=0)
    scale_avg_signif, _ = wavelet.significance(
        var, dt, scales, 2, alpha,
        significance_level=SIG_LVL,
        dof=[scales[sel[0]], scales[sel[-1]]],
        wavelet=mother
    )

    return dict(
        dat_norm=dat_norm, dat_notrend=dat_notrend, std=std, var=var,
        iwave=iwave, wave=wave, scales=scales, freqs=freqs,
        coi=coi, fft=fft, fftfreqs=fftfreqs,
        power=power, fft_power=fft_power, period=period,
        sig95=sig95, fft_theor=fft_theor,
        glbl_power=glbl_power, glbl_signif=glbl_signif,
        scale_avg=scale_avg, scale_avg_signif=scale_avg_signif,
        sel=sel, alpha=alpha,
    )


# =============================================================================
# 3. FIGURE FUNCTIONS
# =============================================================================

def torrence_compo_figure(t, data, cwt_result, title, ylabel, save_path=None):
    """
    Classic 4-panel Torrence & Compo figure:
      a) time series + iwave
      b) wavelet power spectrum
      c) global wavelet spectrum
      d) scale-averaged power
    """
    r  = cwt_result
    dt = t[1] - t[0]

    fig = plt.figure(figsize=(11, 8), dpi=100)

    # --- (a) time series
    ax = fig.add_axes([0.10, 0.75, 0.65, 0.20])
    ax.plot(t, r["iwave"],  "-",  lw=1,   color="0.6",  label="iCWT")
    ax.plot(t, data,        "k",  lw=1.5, label="data")
    ax.set_title(f"a) {title}")
    ax.set_ylabel(ylabel)
    ax.legend(fontsize=8, loc="upper right")

    # --- (b) power spectrum
    bx     = fig.add_axes([0.10, 0.37, 0.65, 0.28], sharex=ax)
    levels = [0.0625, 0.125, 0.25, 0.5, 1, 2, 4, 8, 16]
    bx.contourf(
        t, np.log2(r["period"]), np.log2(r["power"]),
        np.log2(levels), extend="both", cmap="viridis"
    )
    extent = [t.min(), t.max(), 0, r["period"].max()]
    bx.contour(
        t, np.log2(r["period"]), r["sig95"],
        [-99, 1], colors="k", linewidths=1.5, extent=extent
    )
    # cone of influence
    bx.fill(
        np.concatenate([t, t[-1:] + dt, t[-1:] + dt, t[:1] - dt, t[:1] - dt]),
        np.concatenate([
            np.log2(r["coi"]), [1e-9],
            np.log2(r["period"][-1:]),
            np.log2(r["period"][-1:]), [1e-9]
        ]),
        "k", alpha=0.3, hatch="x", label="CoI"
    )
    bx.set_title(f"b) {ylabel} Wavelet Power Spectrum ({MOTHER.name})")
    bx.set_ylabel("Period (years)")
    bx.hlines(
        np.log2(1), t[0], t[-1],
        colors="orange", linestyles="dashed", label="1-yr"
    )
    Yticks = 2 ** np.arange(
        np.ceil(np.log2(r["period"].min())),
        np.ceil(np.log2(r["period"].max()))
    )
    bx.set_yticks(np.log2(Yticks))
    bx.set_yticklabels(Yticks)

    # --- (c) global spectrum
    cx = fig.add_axes([0.77, 0.37, 0.20, 0.28], sharey=bx)
    cx.plot(r["glbl_signif"],                   np.log2(r["period"]), "k--", label="95% sig")
    cx.plot(r["var"] * r["fft_theor"],          np.log2(r["period"]), "--",  color="#cccccc")
    cx.plot(r["var"] * r["fft_power"],          np.log2(1.0 / r["fftfreqs"]),
            "-", color="#cccccc", lw=1)
    cx.plot(r["var"] * r["glbl_power"],         np.log2(r["period"]), "k-",  lw=1.5)
    cx.set_title("c) Global Spectrum")
    cx.set_xlabel(f"Power [{ylabel}²]")
    cx.set_xlim([0, r["glbl_power"].max() + r["var"]])
    cx.set_ylim(np.log2([r["period"].min(), r["period"].max()]))
    cx.set_yticks(np.log2(Yticks))
    cx.set_yticklabels(Yticks)
    plt.setp(cx.get_yticklabels(), visible=False)

    # --- (d) scale-averaged power
    dx = fig.add_axes([0.10, 0.07, 0.65, 0.20], sharex=ax)
    dx.axhline(r["scale_avg_signif"], color="k", ls="--", lw=1)
    dx.plot(t, r["scale_avg"], "k-", lw=1.5)
    dx.set_title(
        f"d) {SCALE_AVG_BAND[0]}–{SCALE_AVG_BAND[1]} yr scale-averaged power"
    )
    dx.set_xlabel("Year")
    dx.set_ylabel("Avg variance")
    ax.set_xlim([t.min(), t.max()])

    plt.suptitle(title, fontsize=12, y=0.99)
    if save_path:
        fig.savefig(save_path, bbox_inches="tight")
        print(f"  Saved → {save_path}")
    plt.show()
    return fig


def overlay_significance_figure(datasets, ylim_years=(0.2, 6), save_path=None):
    """
    Overlay significance contours from multiple CWT results.

    Parameters
    ----------
    datasets : list of dict, each with keys:
        't'      – time array
        'result' – output of run_cwt()
        'label'  – str label
        'color'  – contour colour
        'hatch'  – hatch pattern for filled significance regions
    ylim_years : tuple (min_period, max_period) in years
    """
    fig, ax = plt.subplots(figsize=(14, 4))

    for ds in datasets:
        r   = ds["result"]
        col = ds["color"]
        ax.contour(
            ds["t"], np.log2(r["period"]), r["sig95"],
            [1, 99], colors=col, linewidths=0.8, alpha=0.8, linestyles="solid"
        )
        cs = ax.contourf(
            ds["t"], np.log2(r["period"]), r["sig95"],
            [1, 99], colors="w", hatches=[ds["hatch"]]
        )
        cs.set_edgecolor(col)

    ax.set_ylim(np.log2(ylim_years[0]), np.log2(ylim_years[1]))
    ytix = ax.get_yticks()
    ax.set_yticks(ytix, labels=np.round(2 ** ytix, 2))
    ax.set_ylabel("Period (years)")
    ax.set_xlabel("Year")
    ax.axhline(y=0, color="green", ls="dashed", lw=1, label="1-yr")
    ax.grid(alpha=0.3)

    legend_handles = [
        mpatches.Patch(
            facecolor="white", edgecolor=ds["color"],
            hatch=ds["hatch"], label=ds["label"]
        )
        for ds in datasets
    ]
    ax.legend(
        handles=legend_handles, loc="upper right",
        frameon=True, title="95% sig. regions"
    )
    ax.set_title("Overlaid 95% significance regions")

    if save_path:
        fig.savefig(save_path, bbox_inches="tight")
        print(f"  Saved → {save_path}")
    plt.show()
    return fig


def combined_timeseries_wavelet_figure(
    t_iso, data_iso, cwt_iso,
    datasets_overlay,
    vlines=None,
    xlim=None,
    ylim_isotope=None,
    ylim_period=(0.2, 4),
    save_path=None
):
    """
    Two-panel figure:
      top    – isotope time series with optional trend lines & earthquake markers
      bottom – overlaid significance contours

    Parameters
    ----------
    vlines : list of dict with keys 'x', 'color', 'ls', 'label'
    datasets_overlay : same format as overlay_significance_figure
    """
    fig, axes = plt.subplots(
        2, 1, figsize=(12, 7), sharex=True,
        gridspec_kw={"hspace": 0, "height_ratios": [2, 1]}
    )

    # top panel
    ax = axes[0]
    ax.plot(t_iso, data_iso, "k", lw=1.2, label="data")
    if vlines:
        for vl in vlines:
            ax.axvline(vl["x"], color=vl.get("color", "gray"),
                       ls=vl.get("ls", "-"), lw=1, label=vl.get("label", ""))
    if ylim_isotope:
        ax.set_ylim(ylim_isotope)
    ax.grid(alpha=0.3)
    ax.legend(fontsize=8, loc="upper right")
    ax.set_ylabel(r"$\delta^{13}$C (‰ VPDB)")
    ax.yaxis.tick_right()
    ax.yaxis.set_label_position("right")

    # bottom panel
    bx = axes[1]
    for ds in datasets_overlay:
        r   = ds["result"]
        col = ds["color"]
        bx.contour(
            ds["t"], np.log2(r["period"]), r["sig95"],
            [1, 99], colors=col, linewidths=0.7, alpha=0.8
        )
        cs = bx.contourf(
            ds["t"], np.log2(r["period"]), r["sig95"],
            [1, 99], colors="w", hatches=[ds["hatch"]]
        )
        cs.set_edgecolor(col)

    bx.set_ylim(np.log2(ylim_period[0]), np.log2(ylim_period[1]))
    ytix = bx.get_yticks()
    bx.set_yticks(ytix, labels=np.round(2 ** ytix, 2))
    bx.set_ylabel("Period (years)")
    bx.set_xlabel("Year")
    bx.axhline(y=0, color="green", ls="dashed", lw=1)
    bx.grid(alpha=0.3)

    if vlines:
        for vl in vlines:
            bx.axvline(vl["x"], color=vl.get("color", "gray"),
                       ls=vl.get("ls", "-"), lw=1)

    legend_handles = [
        mpatches.Patch(
            facecolor="white", edgecolor=ds["color"],
            hatch=ds["hatch"], label=ds["label"]
        )
        for ds in datasets_overlay
    ]
    bx.legend(
        handles=legend_handles,
        bbox_to_anchor=(1.01, 1), loc="upper left",
        borderaxespad=0., frameon=False, title="95% sig."
    )

    if xlim:
        axes[0].set_xlim(xlim)

    plt.suptitle(r"$\delta^{13}$C & wavelet significance", fontsize=12)
    if save_path:
        fig.savefig(save_path, bbox_inches="tight")
        print(f"  Saved → {save_path}")
    plt.show()
    return fig


# =============================================================================
# 4. EXTERNAL DATA LOADERS
# =============================================================================

def make_time_array(n, dt, t0):
    """Return decimal-year time array for a series of length n."""
    return np.arange(0, n) * dt + t0


def load_enso(path):
    """Load ENSO index CSV, add decimal_year column."""
    enso = pd.read_csv(path, parse_dates=[0])
    enso["day_of_year"]  = enso["Date"].dt.dayofyear
    enso["total_days"]   = np.where(enso["Date"].dt.is_leap_year, 366, 365)
    enso["decimal_year"] = enso["Date"].dt.year + (enso["day_of_year"] - 1) / enso["total_days"]
    return enso


def load_tide(path):
    """Load tide gauge CSV, resample to monthly, compute anomaly."""
    from datetime import date as _date

    def _dec_to_dt(dec):
        dec = np.asarray(dec)
        year = dec.astype(int)
        dp   = dec - year
        s    = pd.to_datetime([f"{y}-01-01" for y in year])
        e    = pd.to_datetime([f"{y+1}-01-01" for y in year])
        secs = (e - s).total_seconds()
        return s + pd.to_timedelta(dp * secs, unit="s")

    tide = pd.read_csv(path)
    tide["age_dt"] = _dec_to_dt(tide["Date"])
    monthly = (
        tide.set_index("age_dt")[["Sea Level"]]
        .resample("ME").mean()
        .interpolate()
    )
    monthly["sea_level_anom"] = monthly["Sea Level"] - monthly["Sea Level"].mean()
    return monthly


# =============================================================================
# 5. DATA LOADING  –  uses objects produced by 20260223_read_data.py
# =============================================================================
# Run the data-preparation script first (Jupyter / IPython):
#
%run 20260223_read_data.py
#
# In plain Python use:
#   exec(open('20260223_read_data.py').read())
#
# This provides:
#   df_sorted      – raw coral data sorted by age_final, with all MAs and
#                    tectonic-removed columns (d13c_no_tectonic, d18o_pdb_no_tectonic …)
#   df_resampled   – monthly-resampled coral data (resample rule 'ME')
#                    columns: age_resampled, d13c, d18o, d18o_pdb,
#                             *_ma, *_ma2, *_no_tectonic, *_notec_ma, *_cdm
#   coefs_centered – rolling OLS slopes / intercepts
#   dark_bands     – dark-band start / end pairs
#
# External datasets (ENSO, tide gauge) are loaded below.
# Update the paths to match your directory layout.
# =============================================================================

ENSO_PATH = "../../ENSO_Time_Series.csv"
TIDE_PATH = "../../2025-10-10_Fred-TideGauge/memo_edited.csv"


# =============================================================================
# 6. MAIN – run everything
# =============================================================================

if __name__ == "__main__":

    # ------------------------------------------------------------------
    # 5a. Load external datasets
    # ------------------------------------------------------------------
    enso = load_enso(ENSO_PATH)
    tide = load_tide(TIDE_PATH)

    # ------------------------------------------------------------------
    # 5b. Build series & run CWT
    #
    #  Coral t0: first age_resampled value in df_resampled
    #  (no hardcoded 1926.349 – derives directly from the age model)
    # ------------------------------------------------------------------
    print("Running CWT analyses...")

    # --- coral d13C (tectonic-removed, 36-month window)
    coral_d13c_ser = df_resampled["d13c_notec_ma"].dropna()
    T0_CORAL = df_resampled["age_resampled"].dropna().iloc[0]
    t_d13c   = make_time_array(len(coral_d13c_ser), DT, T0_CORAL)
    cwt_d13c = run_cwt(coral_d13c_ser.values)
    print("  ✓ Coral d13C")

    # --- coral d18O PDB (tectonic-removed)
    coral_d18o_ser = df_resampled["d18o_pdb_notec_ma"].dropna()
    t_d18o   = make_time_array(len(coral_d18o_ser), DT, T0_CORAL)
    cwt_d18o = run_cwt(coral_d18o_ser.values)
    print("  ✓ Coral d18O")

    # --- ENSO
    T0_ENSO  = 1950.0
    t_enso   = make_time_array(len(enso), DT, T0_ENSO)
    cwt_enso = run_cwt(enso["ENSO"].values)
    print("  ✓ ENSO")

    # --- tide gauge
    T0_TIDE  = 1968 + 7 / 12
    t_tide   = make_time_array(len(tide), DT, T0_TIDE)
    cwt_tide = run_cwt(tide["sea_level_anom"].values)
    print("  ✓ Tide gauge")

    # ------------------------------------------------------------------
    # 5c. Individual Torrence & Compo figures
    # ------------------------------------------------------------------
    print("\nPlotting individual TC figures...")

    torrence_compo_figure(
        t_d13c, coral_d13c_ser.values, cwt_d13c,
        title="Coral δ¹³C (tectonic-removed)", ylabel="δ¹³C",
        save_path="wavelet_d13c.pdf"
    )
    torrence_compo_figure(
        t_d18o, coral_d18o_ser.values, cwt_d18o,
        title="Coral δ¹⁸O PDB (tectonic-removed)", ylabel="δ¹⁸O",
        save_path="wavelet_d18o.pdf"
    )
    torrence_compo_figure(
        t_enso, enso["ENSO"].values, cwt_enso,
        title="ENSO Index", ylabel="ENSO",
        save_path="wavelet_enso.pdf"
    )
    torrence_compo_figure(
        t_tide, tide["sea_level_anom"].values, cwt_tide,
        title="Honiara Tide Gauge", ylabel="Sea Level (mm)",
        save_path="wavelet_tide.pdf"
    )

    # ------------------------------------------------------------------
    # 5d. Overlay significance figure
    # ------------------------------------------------------------------
    print("\nPlotting overlay significance figure...")

    datasets_overlay = [
        dict(t=t_d13c, result=cwt_d13c, label=r"Coral $\delta^{13}$C", color="k", hatch="\\\\"),
        dict(t=t_d18o, result=cwt_d18o, label=r"Coral $\delta^{18}$O", color="g", hatch="//"),
        dict(t=t_enso, result=cwt_enso, label="ENSO",                   color="r", hatch="||"),
        dict(t=t_tide, result=cwt_tide, label="Tide Gauge",              color="b", hatch="xx"),
    ]

    overlay_significance_figure(
        datasets_overlay,
        ylim_years=(0.2, 6),
        save_path="wavelet_overlay.pdf"
    )

    # ------------------------------------------------------------------
    # 5e. Combined time-series + wavelet panel
    # ------------------------------------------------------------------
    print("\nPlotting combined panel figure...")

    EARTHQUAKE_VLINES = [
        dict(x=1936 + (31 + 29 + 22) / 365, color="#595959", ls="-",  label="M7.1 EQ Apr 1936"),
        dict(x=2007 + (31 + 29 + 31) / 365, color="#595959", ls="--", label="M8.1 EQ Apr 2007"),
    ]

    combined_timeseries_wavelet_figure(
        t_iso=t_d13c,
        data_iso=coral_d13c_ser.values,
        cwt_iso=cwt_d13c,
        datasets_overlay=datasets_overlay,
        vlines=EARTHQUAKE_VLINES,
        xlim=(1927, 1960),
        ylim_isotope=(-2, 0),
        ylim_period=(0.2, 4),
        save_path="wavelet_combined.pdf"
    )

    print("\nAll done.")
```
